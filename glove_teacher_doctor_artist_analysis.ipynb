# GloVe Word Embedding Analizi
Bu notebook, GloVe kelime gömme modeli kullanılarak teacher, doctor ve artist kavramlarının semantik analizini gerçekleştirmek için hazırlanmıştır.

```
from gensim.scripts.glove2word2vec import glove2word2vec

# GloVe ham dosyası
glove_input_file = "/content/glove.6B.50d.txt"  # Kendi dosya yolunuza göre değiştirin
glove_output_file = "/content/glove.6B.50d.word2vec.txt"

# GloVe'ı Word2Vec formatına çevir
glove2word2vec(glove_input_file, glove_output_file)
```

```
from gensim.models import KeyedVectors

# Modeli yükle
model = KeyedVectors.load_word2vec_format(glove_output_file, binary=False)
print("Model başarıyla yüklendi.")
```

```
# Teacher, doctor, artist için en benzer kelimeleri al
print("Teacher kelimesine en benzer kelimeler:")
print(model.most_similar("teacher", topn=5))

print("\nDoctor kelimesine en benzer kelimeler:")
print(model.most_similar("doctor", topn=5))

print("\nArtist kelimesine en benzer kelimeler:")
print(model.most_similar("artist", topn=5))
```

```
# teacher - doctor + artist işlemi
print("teacher - doctor + artist işlemi sonucu:")
result = model.most_similar(positive=["teacher", "artist"], negative=["doctor"], topn=5)
for word, score in result:
    print(f"{word}: {score:.4f}")
```

```
from sklearn.manifold import TSNE
import matplotlib.pyplot as plt
import numpy as np

# Görselleştirilecek kelimeler
words = ['teacher', 'doctor', 'artist', 'student', 'graduate', 'teaching', 'taught', 'teaches',
         'nurse', 'physician', 'patient', 'child',
         'artists', 'musician', 'composer', 'acclaimed', 'filmmaker', 'songwriter', 'performer']

# Vektörleri al
vectors = np.array([model[word] for word in words])

# TSNE ile 2 boyuta indir
tsne = TSNE(n_components=2, perplexity=5, random_state=42, n_iter=1000)
X_tsne = tsne.fit_transform(vectors)

# Görselleştir
plt.figure(figsize=(14, 10))
plt.scatter(X_tsne[:, 0], X_tsne[:, 1])
for i, word in enumerate(words):
    plt.annotate(word, (X_tsne[i, 0], X_tsne[i, 1]))
plt.title("Teacher, Doctor, Artist ve İlişkili Kavramlar - TSNE Görselleştirme")
plt.grid(True)
plt.show()
```
